```python
# Importing essential libraries for data analysis
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Loading the dataset into a pandas DataFrame
# Note: If your extracted file is just named 'train', change 'train.csv' to 'train'
df = pd.read_csv('train.csv')

# Displaying the first 5 rows of the dataset
print("--- First 5 Rows of the Dataset ---")
display(df.head())

# Checking basic information (rows, columns, null values, and data types)
print("\n--- Dataset Information ---")
df.info()
```

```python
# Checking for missing (null) values in each column
print("--- Missing Values Count ---")
display(df.isnull().sum())

# Generating descriptive statistics for numerical columns (count, mean, std, min, max, etc.)
print("\n--- Descriptive Statistics ---")
display(df.describe())

# Checking the mode (most frequent value) for categorical/text columns
print("\n--- Mode of Categorical Columns ---")
display(df.mode().iloc[0])
```

```python
# 1. Converting 'Order Date' to proper datetime format
# (dayfirst=True handles dates like DD/MM/YYYY correctly)
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)

# 2. Extracting Month and Year for time series grouping
df['YearMonth'] = df['Order Date'].dt.to_period('M')

# 3. Grouping data to get total sales per month
monthly_sales = df.groupby('YearMonth')['Sales'].sum().reset_index()
monthly_sales['YearMonth'] = monthly_sales['YearMonth'].astype(str) # Converting to string for plotting

# 4. Plotting the Monthly Sales Trend
plt.figure(figsize=(15, 6))
sns.lineplot(data=monthly_sales, x='YearMonth', y='Sales', marker='o', color='teal', linewidth=2)

# Formatting the chart to make it look professional
plt.title('Monthly Sales Trend', fontsize=16, fontweight='bold')
plt.xlabel('Month-Year', fontsize=12)
plt.ylabel('Total Sales ($)', fontsize=12)
plt.xticks(rotation=45) # Rotating x-axis labels so they don't overlap
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Setting up the layout for 2 side-by-side charts
plt.figure(figsize=(16, 6))

# --- Chart 1: Revenue by Product Category ---
plt.subplot(1, 2, 1) # (1 row, 2 columns, 1st plot)
category_sales = df.groupby('Category')['Sales'].sum().reset_index()
sns.barplot(data=category_sales, x='Category', y='Sales', palette='viridis')
plt.title('Total Revenue by Product Category', fontsize=14, fontweight='bold')
plt.ylabel('Total Sales ($)')
plt.xlabel('Category')

# --- Chart 2: Top 10 Best-Selling Products ---
plt.subplot(1, 2, 2) # (1 row, 2 columns, 2nd plot)
top_products = df.groupby('Product Name')['Sales'].sum().nlargest(10).reset_index()
sns.barplot(data=top_products, y='Product Name', x='Sales', palette='magma')
plt.title('Top 10 Best-Selling Products by Revenue', fontsize=14, fontweight='bold')
plt.xlabel('Total Sales ($)')
plt.ylabel('')

# Adjusting layout so things don't overlap
plt.tight_layout()
plt.show()
```

```python
# Setting up the layout for 2 charts
plt.figure(figsize=(15, 6))

# --- Chart 1: Sales by Customer Segment (Pie Chart) ---
plt.subplot(1, 2, 1)
segment_sales = df.groupby('Segment')['Sales'].sum()
# Using a pie chart to see the percentage share of each segment
plt.pie(segment_sales, labels=segment_sales.index, autopct='%1.1f%%',
        startangle=90, colors=['#ff9999','#66b3ff','#99ff99'], wedgeprops={'edgecolor': 'black'})
plt.title('Revenue Share by Customer Segment', fontsize=14, fontweight='bold')

# --- Chart 2: Correlation Heatmap ---
plt.subplot(1, 2, 2)
# Selecting only numerical columns for correlation (ignoring IDs like Row ID and Postal Code)
numerical_df = df.select_dtypes(include=['float64', 'int64']).drop(columns=['Row ID', 'Postal Code'], errors='ignore')

# Creating the heatmap
sns.heatmap(numerical_df.corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap (Numerical Data)', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# --- Additional Insight: Sales by Region and Category ---
plt.figure(figsize=(12, 6))

# Grouped bar chart using seaborn
sns.barplot(data=df, x='Region', y='Sales', hue='Category', estimator=sum, errorbar=None, palette='Set2')

plt.title('Total Sales by Region and Category (Non-Obvious Insight)', fontsize=14, fontweight='bold')
plt.ylabel('Total Sales ($)')
plt.xlabel('Region')
plt.legend(title='Category')

plt.tight_layout()
plt.show()
```
