```python
import pandas as pd
import numpy as np

# 1. Loading the messy dataset
df = pd.read_csv('tested.csv')

# 2. Displaying the first 5 rows
display(df.head())

# 3. Checking for missing (null) values in each column
print("\n--- Missing Values Count ---")
print(df.isnull().sum())

# 4. Checking data types and total rows
print("\n--- Data Information ---")
df.info()
```

```python
# 1. Fill missing 'Age' values with the mean (average) age
df['Age'] = df['Age'].fillna(df['Age'].mean())

# 2. Fill the 1 missing 'Fare' value with the mean fare
df['Fare'] = df['Fare'].fillna(df['Fare'].mean())

# 3. Drop the 'Cabin' column because it has too many missing values (327 out of 418)
df = df.drop(columns=['Cabin'])

# 4. Verify that no missing values are left
print("--- Missing Values After Cleaning ---")
print(df.isnull().sum())
```

```python
# 5. Save the cleaned dataset to a new CSV file
df.to_csv('cleaned_tested.csv', index=False)

print("✅ Cleaned data saved successfully as 'cleaned_tested.csv'!")
```
